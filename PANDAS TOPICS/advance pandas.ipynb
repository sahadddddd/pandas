```python
# What are Outliers?

# . Values far away from normal data

# . Can affect analysis and ML models

# . May be errors or rare observations

# · [10, 12, 15, 18, 200]
```

```python
# Detecting & Removing Outliers 

# Common Methods:
# · Box Plot
# . IQR Method
# · Z-Score Method
```

```python
#using box plot
import matplotlib.pyplot as plt
plt.boxplot(df["Age"])
```

```python
import matplotlib.pyplot as plt
plt.boxplot(df["Fare"])
```

```python
# using IQR methord

q1=df["Age"].quantile(0.25)
q3=df["Age"].quantile(0.75)
iqr=q3-q1
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr
outliers=df[(df["Age"]<lower_bound)|(df["Age"]>upper_bound)]
print(outliers)
```

```python
#USING ZCORE METHORD
mean=df["Age"].mean()
std=df["Age"].std()
df["z_score"]=(df["Age"]-mean)/std
outliers=df[df["z_score"].abs()>3]
outliers[["Age", "z_score"]]
```

```python
print(outliers["Age"])
print("Number of outliers:", len(outliers))#To count how many outliers were found

# To see the highest Z-scores:
print(outliers[["Age", "z_score"]].sort_values(
    by="z_score",
    ascending=False
))
```

```python
# Normalization (Min-Max Scaling)
# . Rescales features to a fixed range, usually between O and 1.
# . Drawback: Highly sensitive to outliers, which can compress most of the data into a tiny range.

# Formula: X_scaled=(X-Xmin)/(Xmax-Xmin)
```

```python
# Standardization (Z-score Normalization)
# . Transforms data so it has a mean (u) of O and a standard deviation (o) of l.
# . Less sensitive to outliers, when compared to Normalization.

# Formula: X_scaled=(x-mean)/std
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Ali", "Sara", "Sara"],
    "Subject": ["Math", "Science", "Math", "Science"],
    "Marks": [80, 90, 85, 95]
})
print(df)
```

```python
#EXAMPLE FOR PIVOT
pivot_df = df.pivot(
    index="Name",
    columns="Subject",
    values="Marks"
)
print(pivot_df)
```

```python
#EXAMPLE FOR MELT
melt_df = pivot_df.reset_index().melt(
    id_vars="Name",
    var_name="Subject",
    value_name="Marks"
)
print(melt_df)
```

```python
# TWO METHORDS OF FEAUTURE SCALING

# 1.MinMax normalization
# 2.standardization using z-score
```

```python
# 1.MinMax normalization
# Min-Max Normalization transforms numerical values into a specified range, usually between 0 and 1.

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df["Age_Scaled"] = scaler.fit_transform(df[["Age"]])
# df["Age_Scaled"] = scaler.fit_transform(df[["Age"]]).ravel()-we use .ravel to get one dimentsional
df.head(6)
```

```python
# 2.standardization using z-score
# Unlike Min-Max Normalization, standardization does not restrict every value to the range 0 to 1.
# A standardized value can be negative, zero, or positive.
from sklearn.preprocessing import StandardScaler
Scaler=StandardScaler()
df["Age standardized"]=Scaler.fit_transform(df[["Age"]]).ravel()
df
```
